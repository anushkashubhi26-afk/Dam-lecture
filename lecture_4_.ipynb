```python
# !pip install seaborn plotly openpyxl

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_excel("/content/Zomato_India_Orders_Dataset.xlsx" , sheet_name="Orders_Data")
print("Data loaded:", df.shape [0], "orders and ", df.shape [1], "variable")
```

```python
# 1. Shape: tells us the number of rows and columns
print("Shape:", df.shape)

# 2. Columns: shows the names of all columns
print("/nColumns:")
print(df.columns)

# 3. Index: shows the row labels
print("\nIndex:")
print(df.index)

# 4. Data Types: tells us the type of data in each column
print("\nData Types:")
print(df.dtypes)


```

```python
# 1.5 Classify the variables (decide this using the data dictionary)

number_cols = [
    "Items_Count",
    "Order_Value",
    "Discount_Pct",
    "Delivery_Fee",
    "Distance_Km",
    "Delivery_Time_Min",
    "Rating",
    "Final_Amount"
]

category_cols = [
    "City",
    "Cuisine",
    "Customer_Segment",
    "Platform",
    "Payment_Mode"
]

date_cols = ["Order_Date"]

id_cols = ["Order_ID"]

print("Numerical :", len(number_cols), number_cols)
print("Categorical:", len(category_cols), category_cols)
print("Date      :", date_cols)
print("ID        :", id_cols)
```

```python
# 1.6 Data quality check: missing values and duplicates

print("Missing values in each column:")
print(df.isnull().sum())

print("\nColumns with missing data:")
missing = df.isnull().sum()
print(missing[missing > 0])

print(
    "Percent missing in Rating:",
    round(df["Rating"].isnull().mean() * 100, 1),
    "%"
)

print("\nDuplicate rows:", df.duplicated().sum())

print("Is Order ID unique?:", df["Order_ID"].is_unique)
```

```python
# 2.1 describe(): one command, many statistics (numerical columns)
df[number_cols].describe().round(2)
```

```python
# 2.2 Statitics for one important column: Final_Amount (what the customer pay)
amount = df["Final_Amount"]

print("Mean (average)       : Rs", round(amount.mean(), 2))
print("Median (middle)      : Rs", amount.median())
print("Mode (most common)   : Rs", amount.mode()[0])
print("Minimum              : Rs", amount.min())
print("Maximum              : Rs", amount.max())
print("Standard deviation   : Rs", round(amount.std(), 2))
print("Range (max - min)    : Rs", amount.max() - amount.min())
```
